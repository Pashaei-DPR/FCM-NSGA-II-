```python
# %%
# ==============================================================================
# 1. IMPORTS & GLOBAL SETTINGS
# ==============================================================================
import warnings
from itertools import product
from scipy import stats
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
import seaborn as sns
import skfuzzy as fuzz

# Safe display() import for running as a standard .py script or in Jupyter
try:
    from IPython.display import display
except ImportError:
    def display(obj):
        print(obj)

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, cohen_kappa_score
from sklearn.decomposition import PCA

from pymoo.core.problem import ElementwiseProblem
from pymoo.core.callback import Callback
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.operators.crossover.sbx import SBX
from pymoo.operators.mutation.pm import PM
from pymoo.operators.sampling.rnd import FloatRandomSampling
from pymoo.indicators.hv import HV

warnings.filterwarnings("ignore")
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)


# %%
# ==============================================================================
# 2. DATA LOADING, PREPROCESSING & CONSTANTS
# ==============================================================================
RANDOM_SEEDS = [42, 123, 1234, 12345, 123456]
PRIMARY_SEED = RANDOM_SEEDS[0]
CLUSTER_COUNTS = range(2, 7)

feature_columns = ['ELBOW_ANGLE', 'SHOULDER_ANGLE', 'HIP_ANGLE', 'KNEE_ANGLE']
short = [f.split('_')[0].title() for f in feature_columns]

TARGET_ANGLES = np.array([90, 90, 180, 180])
K_STD = 2.0
MAX_ERRORS = 2
TOL = 14.0
RULE_NAME = f'Rule-based (±{TOL:.0f}°)'

df = pd.read_csv('plank_classification.csv')
X = df[feature_columns].apply(pd.to_numeric, errors='raise')
raw_angles = X.to_numpy(float)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"Loaded {len(df)} training samples.")


# %%
# ==============================================================================
# 3. K-MEANS CLUSTERING (TRAIN)
# ==============================================================================
kmeans_scores = {}
for k in CLUSTER_COUNTS:
    scores = []
    for seed in RANDOM_SEEDS:
        model = KMeans(n_clusters=k, random_state=seed, n_init=20)
        labels = model.fit_predict(X_scaled)
        scores.append(silhouette_score(X_scaled, labels))
    kmeans_scores[k] = (float(np.mean(scores)), float(np.std(scores, ddof=1)))

best_kmeans_k = max(kmeans_scores, key=lambda k: (kmeans_scores[k][0], -k))
best_kmeans_score, best_kmeans_sd = kmeans_scores[best_kmeans_k]

final_kmeans = KMeans(n_clusters=best_kmeans_k, random_state=PRIMARY_SEED, n_init=20)
best_kmeans_labels = final_kmeans.fit_predict(X_scaled)

print('K-Means mean silhouette by k (mean ± SD across seeds):', kmeans_scores)
print(f'K-Means selected k={best_kmeans_k}; mean silhouette={best_kmeans_score:.4f} ± {best_kmeans_sd:.4f}; final fit seed={PRIMARY_SEED}')
df['kmeans_cluster'] = best_kmeans_labels


# %%
# ==============================================================================
# 4. DBSCAN CLUSTERING TUNED WITH NSGA-II (TRAIN)
# ==============================================================================
class DBSCANProblem(ElementwiseProblem):
    def __init__(self, X_data):
        self.X = X_data
        # decision variables: [eps, min_samples]
        super().__init__(n_var=2, n_obj=2,
                         xl=np.array([0.1, 3]),
                         xu=np.array([2.0, 15]))

    def _evaluate(self, x, out, *args, **kwargs):
        eps = float(x[0])
        min_samples = int(round(x[1]))

        labels = DBSCAN(eps=eps, min_samples=min_samples).fit_predict(self.X)
        mask = labels != -1
        noise_ratio = 1.0 - mask.mean()

        if len(set(labels[mask])) < 2:
            out["F"] = [1.0, 1.0]
            return

        sil = silhouette_score(self.X[mask], labels[mask])
        out["F"] = [-sil, noise_ratio]   # maximize silhouette, minimize noise


problem_db = DBSCANProblem(X_scaled)
all_F_db, all_X_db, all_seeds_db = [], [], []

for seed in RANDOM_SEEDS:
    algorithm = NSGA2(
        pop_size=40,
        sampling=FloatRandomSampling(),
        crossover=SBX(prob=0.9, eta=15),
        mutation=PM(eta=20),
        eliminate_duplicates=True
    )
    res = minimize(problem_db, algorithm, ('n_gen', 50), seed=seed, verbose=False)

    all_F_db.append(res.F)
    all_X_db.append(res.X)
    all_seeds_db.extend([seed] * len(res.F))
    print(f"Seed {seed}: {len(res.F)} Pareto solutions")

F_db = np.vstack(all_F_db)
X_db = np.vstack(all_X_db)
all_seeds_db = np.array(all_seeds_db)

# Utopia Point selection on the pooled fronts
F_db_norm = (F_db - F_db.min(axis=0)) / (F_db.max(axis=0) - F_db.min(axis=0) + 1e-12)
best_db_idx = np.argmin(np.linalg.norm(F_db_norm, axis=1))

best_db_eps = float(X_db[best_db_idx][0])
best_db_min_samples = int(round(X_db[best_db_idx][1]))
best_db_seed = int(all_seeds_db[best_db_idx])
best_db_score = float(-F_db[best_db_idx][0])
best_db_noise = float(F_db[best_db_idx][1])

print(f"DBSCAN (NSGA-II) Best Params -> eps: {best_db_eps:.3f} | "
      f"min_samples: {best_db_min_samples} | Seed: {best_db_seed} | "
      f"Silhouette: {best_db_score:.4f} | Noise: {best_db_noise:.2%}")

df['dbscan_cluster'] = DBSCAN(eps=best_db_eps, min_samples=best_db_min_samples).fit_predict(X_scaled)
print(df['dbscan_cluster'].value_counts().sort_index())


# %%
# ==============================================================================
# 5. HIERARCHICAL (AGGLOMERATIVE) CLUSTERING (TRAIN)
# ==============================================================================
agg_scores = {}
for k in CLUSTER_COUNTS:
    labels = AgglomerativeClustering(n_clusters=k, linkage='ward').fit_predict(X_scaled)
    agg_scores[k] = silhouette_score(X_scaled, labels)

best_agg_k = max(agg_scores, key=lambda k: (agg_scores[k], -k))
best_agg_score = agg_scores[best_agg_k]
best_agg_labels = AgglomerativeClustering(n_clusters=best_agg_k, linkage='ward').fit_predict(X_scaled)

print('Hierarchical silhouette by k:', agg_scores)
print(f'Hierarchical selected k={best_agg_k}; silhouette={best_agg_score:.4f}')
df['hierarchical_cluster'] = best_agg_labels


# %%
# ==============================================================================
# 6. GAUSSIAN MIXTURE MODEL (GMM) CLUSTERING (TRAIN)
# ==============================================================================
gmm_scores = {}
for k in CLUSTER_COUNTS:
    scores = []
    for seed in RANDOM_SEEDS:
        model = GaussianMixture(n_components=k, random_state=seed, n_init=5, reg_covar=1e-6)
        labels = model.fit_predict(X_scaled)
        scores.append(silhouette_score(X_scaled, labels))
    gmm_scores[k] = (float(np.mean(scores)), float(np.std(scores, ddof=1)))

best_gmm_k = max(gmm_scores, key=lambda k: (gmm_scores[k][0], -k))
best_gmm_score, best_gmm_sd = gmm_scores[best_gmm_k]

final_gmm = GaussianMixture(n_components=best_gmm_k, random_state=PRIMARY_SEED, n_init=5, reg_covar=1e-6)
best_gmm_labels = final_gmm.fit_predict(X_scaled)

print('GMM mean silhouette by k (mean ± SD across seeds):', gmm_scores)
print(f'GMM selected components={best_gmm_k}; mean silhouette={best_gmm_score:.4f} ± {best_gmm_sd:.4f}; final fit seed={PRIMARY_SEED}')
df['gmm_cluster'] = best_gmm_labels


# %%
# ==============================================================================
# 7. FUZZY C-MEANS (FCM) CLUSTERING (TRAIN)
# ==============================================================================
fcm_data = X_scaled.T  # skfuzzy expects (features, samples)

def fcm_membership(Xs, centers, m):
    d = np.linalg.norm(Xs[:, None, :] - centers[None, :, :], axis=2) + 1e-12
    w = d ** (-2.0 / (m - 1.0))
    return w / w.sum(1, keepdims=True)  # shape: (samples, clusters)

FUZZINESS = [1.5, 2.0, 2.5]

fcm_scores = {}
for c in CLUSTER_COUNTS:
    for m in FUZZINESS:
        fpcs = []
        for seed in RANDOM_SEEDS:
            *_, fpc = fuzz.cluster.cmeans(fcm_data, c=c, m=m, error=0.005, maxiter=1000, init=None, seed=seed)
            fpcs.append(float(fpc))
        fcm_scores[(c, m)] = (float(np.mean(fpcs)), float(np.std(fpcs, ddof=1)))

best_fcm_c, best_fcm_m = max(fcm_scores, key=lambda p: (fcm_scores[p][0], -p[0], -p[1]))
best_fcm_score, best_fcm_sd = fcm_scores[(best_fcm_c, best_fcm_m)]

final_fcm_centers, best_fcm_u, *_, final_fcm_fpc = fuzz.cluster.cmeans(
    fcm_data, c=best_fcm_c, m=best_fcm_m, error=0.005, maxiter=1000, init=None, seed=PRIMARY_SEED
)
train_mem = fcm_membership(X_scaled, final_fcm_centers, best_fcm_m)
best_fcm_labels = train_mem.argmax(1)

print(f"Label agreement (skfuzzy u vs recomputed, train): {(best_fcm_u.argmax(0) == best_fcm_labels).mean():.2%}")
print('FCM mean FPC by (c, m) (mean ± SD across seeds):', fcm_scores)
print(f'FCM selected c={best_fcm_c}, m={best_fcm_m}; mean FPC={best_fcm_score:.4f} ± {best_fcm_sd:.4f}; final fit FPC={final_fcm_fpc:.4f}; seed={PRIMARY_SEED}')

df['fcm_cluster'] = best_fcm_labels
for i in range(best_fcm_c):
    df[f'Prob_Cluster_{i}'] = np.round(train_mem[:, i] * 100, 2)


# %%
# ==============================================================================
# 8. FCM OPTIMIZED MEMBERSHIP THRESHOLD (NSGA-II)
# ==============================================================================
# "Correct posture" cluster = FCM cluster whose raw-angle mean is nearest the ideal posture
fcm_means = df.groupby('fcm_cluster')[feature_columns].mean()
fcm_target = int(fcm_means.index[np.linalg.norm(fcm_means.values - TARGET_ANGLES, axis=1).argmin()])
in_target = (df['fcm_cluster'].to_numpy() == fcm_target)
target_prob = df[f'Prob_Cluster_{fcm_target}'].to_numpy()


class ThresholdProblem(ElementwiseProblem):
    def __init__(self):
        # decision variable: membership threshold (%)
        super().__init__(n_var=1, n_obj=2, xl=np.array([50.0]), xu=np.array([99.9]))

    def _evaluate(self, x, out, *args, **kwargs):
        mask = in_target & (target_prob >= x[0])
        if mask.sum() == 0:
            out["F"] = [1.0, 10000.0]
            return
        retention_cost = 1.0 - mask.sum() / len(df)                              # minimize dropped samples
        angle_error = np.mean((raw_angles[mask].mean(0) - TARGET_ANGLES) ** 2)   # minimize angle MSE
        out["F"] = [retention_cost, angle_error]


problem_thr = ThresholdProblem()
all_F, all_X, all_seeds = [], [], []

for seed in RANDOM_SEEDS:
    algorithm = NSGA2(
        pop_size=40,
        n_offsprings=10,
        sampling=FloatRandomSampling(),
        crossover=SBX(prob=0.9, eta=15),
        mutation=PM(eta=20),
        eliminate_duplicates=True
    )
    res_opt = minimize(problem_thr, algorithm, ('n_gen', 50), seed=seed, verbose=False)
    all_F.append(res_opt.F)
    all_X.append(res_opt.X)
    all_seeds.extend([seed] * len(res_opt.F))
    print(f"Seed {seed}: {len(res_opt.F)} Pareto solutions")

F_opt = np.vstack(all_F)
X_opt = np.vstack(all_X)
all_seeds = np.array(all_seeds)

# Utopia / knee point selection on the pooled fronts
F_norm = (F_opt - F_opt.min(axis=0)) / (F_opt.max(axis=0) - F_opt.min(axis=0) + 1e-12)
best_idx = np.argmin(np.linalg.norm(F_norm, axis=1))

best_fcm_threshold = float(X_opt[best_idx][0])
best_th_seed = int(all_seeds[best_idx])
best_th_retention = float(F_opt[best_idx][0])
best_th_angle_error = float(F_opt[best_idx][1])

print(f"FCM threshold (NSGA-II) -> threshold: {best_fcm_threshold:.2f}% | Seed: {best_th_seed} | "
      f"Retention cost: {best_th_retention:.4f} | Angle MSE: {best_th_angle_error:.2f}")

FCM_DISCARD_ID = int(df['fcm_cluster'].max()) + 1
dropped = in_target & (target_prob < best_fcm_threshold)
df['fcm_opt_cluster'] = np.where(dropped, FCM_DISCARD_ID, df['fcm_cluster'])

print(f"Correct-posture cluster: {fcm_target} | discarded cluster id: {FCM_DISCARD_ID} | moved: {int(dropped.sum())}")
print(df['fcm_opt_cluster'].value_counts().sort_index())


# %%
# ==============================================================================
# 9. PARETO FRONT DIAGRAMS (STANDARD & PUBLICATION-QUALITY)
# ==============================================================================
utopia = F_opt.min(axis=0)

def pareto_mask(F):
    keep = np.ones(len(F), dtype=bool)
    for i in range(len(F)):
        dominated = np.all(F <= F[i], axis=1) & np.any(F < F[i], axis=1)
        keep[i] = not np.any(dominated)
    return keep

pm = pareto_mask(F_opt)
F_front, seeds_front = F_opt[pm], all_seeds[pm]
order_idx = np.argsort(F_front[:, 0])
F_line = F_front[order_idx]

knee_pt = F_opt[best_idx]
i_ret = np.argmin(F_opt[:, 0])
i_err = np.argmin(F_opt[:, 1])

# ---- Standard Plot ----
fig, ax = plt.subplots(figsize=(9, 6), dpi=200)
ax.scatter(F_opt[~pm, 0], F_opt[~pm, 1], s=25, c='lightgray', label='Dominated (pooled)')
for s in RANDOM_SEEDS:
    m = seeds_front == s
    ax.scatter(F_front[m, 0], F_front[m, 1], s=45, alpha=0.7, label=f'Pareto front (seed {s})')

ax.plot(F_line[:, 0], F_line[:, 1], 'k--', lw=1, alpha=0.5, label='Pooled Pareto front')
ax.scatter(*utopia, marker='X', s=220, c='black', edgecolors='white', zorder=10, label='Utopia point')
ax.scatter(*F_opt[i_ret], marker='o', s=160, c='green', edgecolors='k', zorder=9, label='Best retention')
ax.scatter(*F_opt[i_err], marker='o', s=160, c='purple', edgecolors='k', zorder=9, label='Best angle error')
ax.scatter(*knee_pt, marker='*', s=400, c='red', edgecolors='k', zorder=11,
           label=f'Knee point (threshold {best_fcm_threshold:.1f}%)')
ax.plot([utopia[0], knee_pt[0]], [utopia[1], knee_pt[1]], c='red', ls=':', lw=1.5)

ax.set_xlabel('Objective 1: Retention cost ')
ax.set_ylabel('Objective 2: Angle error, MSE ')
ax.set_title('Pareto front of threshold optimization (NSGA-II, FCM)')
ax.grid(True, ls='--', alpha=0.6)
ax.legend(fontsize=8, loc='best')
plt.tight_layout()
plt.savefig('pareto_front_utopia.png', dpi=300, bbox_inches='tight')
plt.show()

# ---- Publication-Quality Plot ----
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 10,
    "axes.labelsize": 11,
    "axes.titlesize": 12,
    "legend.fontsize": 8,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "axes.linewidth": 0.8,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

seed_colors = {
    42:     "#0072B2",  # Blue
    123:    "#E69F00",  # Orange
    1234:   "#009E73",  # Green
    12345:  "#CC79A7",  # Pink
    123456: "#D55E00",  # Vermilion
}

fig, ax = plt.subplots(figsize=(7.2, 4.8))
ax.scatter(F_opt[~pm, 0], F_opt[~pm, 1], s=20, color="lightgray", alpha=0.65,
           edgecolors="none", label="Dominated solutions", zorder=1)

for s in RANDOM_SEEDS:
    m = seeds_front == s
    if np.any(m):
        ax.scatter(F_front[m, 0], F_front[m, 1], s=38, color=seed_colors.get(s, "steelblue"),
                   alpha=0.85, edgecolors="white", linewidths=0.5, label=f"Seed {s}", zorder=3)

ax.plot(F_line[:, 0], F_line[:, 1], color="black", linewidth=1.3, linestyle="--",
        label="Pooled Pareto front", zorder=2)
ax.plot([utopia[0], knee_pt[0]], [utopia[1], knee_pt[1]], color="red", linestyle=":",
        linewidth=1.5, label="Utopia–knee distance", zorder=4)
ax.scatter(*utopia, marker="X", s=110, color="black", edgecolors="white",
           linewidths=0.7, label="Utopia point", zorder=6)
ax.scatter(*F_opt[i_ret], marker="o", s=75, color="limegreen", edgecolors="black",
           linewidths=0.7, label="Best retention", zorder=5)
ax.scatter(*F_opt[i_err], marker="o", s=75, color="mediumpurple", edgecolors="black",
           linewidths=0.7, label="Best angle error", zorder=5)
ax.scatter(*knee_pt, marker="*", s=230, color="red", edgecolors="black",
           linewidths=0.7, label=f"Knee point ({best_fcm_threshold:.1f}%)", zorder=7)

ax.set_xlabel("Retention cost ")
ax.set_ylabel("Angular error, MSE ")
ax.grid(True, linestyle=":", linewidth=0.6, color="0.85")
ax.set_axisbelow(True)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.tick_params(direction="out", length=3, width=0.8)

handles, labels_leg = ax.get_legend_handles_labels()
unique = dict(zip(labels_leg, handles))
ax.legend(unique.values(), unique.keys(), frameon=False, loc="best", ncol=2,
          handletextpad=0.5, columnspacing=1.0)

fig.tight_layout()
fig.savefig("pareto_front.pdf", bbox_inches="tight")
fig.savefig("pareto_front.png", dpi=600, bbox_inches="tight")
fig.savefig("pareto_front.svg", bbox_inches="tight")
plt.show()


# %%
# ==============================================================================
# 10. FCM THRESHOLD GRID SEARCH & HYPERVOLUME COMPARISON VS. NSGA-II
# ==============================================================================
GRID = np.unique(np.append(np.arange(50.0, 99.9, 1.0), 99.9))

def threshold_objectives(t):
    mask = in_target & (target_prob >= t)
    if mask.sum() == 0:
        return np.array([1.0, 10000.0])
    retention_cost = 1.0 - mask.sum() / len(df)
    angle_error = np.mean((raw_angles[mask].mean(0) - TARGET_ANGLES) ** 2)
    return np.array([retention_cost, angle_error])

F_grid_all = np.array([threshold_objectives(t) for t in GRID])

pm_grid = pareto_mask(F_grid_all)
F_grid, T_grid = F_grid_all[pm_grid], GRID[pm_grid]
_, ui = np.unique(F_grid, axis=0, return_index=True)
F_grid, T_grid = F_grid[ui], T_grid[ui]

F_grid_norm = (F_grid - F_grid.min(0)) / (F_grid.max(0) - F_grid.min(0) + 1e-12)
grid_idx = np.argmin(np.linalg.norm(F_grid_norm, axis=1))
best_grid_threshold = float(T_grid[grid_idx])

print(f"Grid search -> threshold: {best_grid_threshold:.2f}% | "
      f"Retention cost: {F_grid[grid_idx][0]:.4f} | Angle MSE: {F_grid[grid_idx][1]:.2f} | "
      f"Evaluations: {len(GRID)}")

dropped_grid = in_target & (target_prob < best_grid_threshold)
df['fcm_grid_cluster'] = np.where(dropped_grid, FCM_DISCARD_ID, df['fcm_cluster'])
print(f"Moved to discarded cluster: {int(dropped_grid.sum())}")
print(df['fcm_grid_cluster'].value_counts().sort_index())

# Hypervolume comparison table (saved to CSV)
F_nsga = F_opt[pareto_mask(F_opt)]
allF = np.vstack([F_nsga, F_grid])
mn, mx = allF.min(0), allF.max(0)
norm_fn = lambda F: (F - mn) / np.where(mx - mn > 0, mx - mn, 1e-9)
hv = HV(ref_point=np.array([1.1, 1.1]))

cmp_tbl = pd.DataFrame({
    'Method': ['NSGA-II', 'Grid search (1% step)'],
    'Evaluations': [len(RANDOM_SEEDS) * 530, len(GRID)],
    'Front size': [len(F_nsga), len(F_grid)],
    'Hypervolume (normalised, higher=better)': [hv(norm_fn(F_nsga)), hv(norm_fn(F_grid))],
    'Chosen threshold (%)': [best_fcm_threshold, best_grid_threshold],
}).round(4)
display(cmp_tbl)
cmp_tbl.to_csv('table_nsga_vs_grid_hv.csv', index=False)

# Standard comparison plot
fig, ax = plt.subplots(figsize=(9, 6), dpi=200)
ax.scatter(F_grid[:, 0], F_grid[:, 1], s=30, c='tab:orange', label='Grid-search front')
ax.scatter(F_nsga[:, 0], F_nsga[:, 1], s=60, facecolors='none', edgecolors='tab:blue', label='NSGA-II front')
ax.scatter(*F_grid[grid_idx], marker='*', s=350, c='tab:orange', edgecolors='k', zorder=10,
           label=f'Grid knee ({best_grid_threshold:.1f}%)')
ax.scatter(*F_opt[best_idx], marker='*', s=350, c='tab:blue', edgecolors='k', zorder=10,
           label=f'NSGA-II knee ({best_fcm_threshold:.1f}%)')
ax.scatter(*utopia, marker='X', s=200, c='black', edgecolors='white', zorder=11, label='Utopia point')
ax.set_xlabel('Retention cost ')
ax.set_ylabel('Angle error, MSE ')
ax.set_title('Pareto fronts: NSGA-II vs grid search')
ax.grid(True, ls='--', alpha=0.6)
ax.legend(fontsize=8)
plt.tight_layout()
plt.savefig('pareto_nsga_vs_grid.png', dpi=300, bbox_inches='tight')
plt.show()

# Publication-ready comparison plot
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "DejaVu Serif", "Times"],
    "font.size": 11,
    "axes.labelsize": 12,
    "axes.titlesize": 13,
    "legend.fontsize": 9,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "axes.linewidth": 0.8,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
    "lines.linewidth": 1.2,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

fig, ax = plt.subplots(figsize=(5.5, 4.0), dpi=300)
c_grid, c_nsga, c_utopia = "#E69F00", "#0072B2", "#000000"

ax.scatter(F_grid[:, 0], F_grid[:, 1], s=28, c=c_grid, alpha=0.85, edgecolors="none",
           zorder=3, label="Grid search front")
ax.scatter(F_nsga[:, 0], F_nsga[:, 1], s=55, facecolors="none", edgecolors=c_nsga,
           linewidths=1.4, zorder=4, label="NSGA-II front")
ax.scatter(*F_grid[grid_idx], marker="*", s=220, c=c_grid, edgecolors="k",
           linewidths=0.6, zorder=6, label=f"Grid knee ({best_grid_threshold:.1f} %)")
ax.scatter(*F_opt[best_idx], marker="*", s=220, c=c_nsga, edgecolors="k",
           linewidths=0.6, zorder=6, label=f"NSGA-II knee ({best_fcm_threshold:.1f} %)")
ax.scatter(*utopia, marker="X", s=110, c=c_utopia, edgecolors="white",
           linewidths=0.8, zorder=7, label="Utopia")

ax.set_xlabel("Retention cost ")
ax.set_ylabel("Angle error MSE ")
ax.set_title("Pareto fronts: NSGA-II vs. grid search", pad=8)
ax.grid(True, which="major", ls="--", lw=0.5, alpha=0.45, zorder=0)
ax.set_axisbelow(True)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax.legend(loc="upper right", frameon=True, fancybox=False,
          edgecolor="0.7", framealpha=0.95, borderpad=0.4)
plt.tight_layout(pad=0.3)
fig.savefig("pareto_nsga_vs_grid.png", dpi=600, bbox_inches="tight", facecolor="white", edgecolor="none")
fig.savefig("pareto_nsga_vs_grid.pdf", bbox_inches="tight", facecolor="white", edgecolor="none")
fig.savefig("pareto_nsga_vs_grid.svg", bbox_inches="tight", facecolor="white", edgecolor="none")
plt.show()


# %%
# ==============================================================================
# 11. matched retention-RETENTION DBSCAN & RULE-BASED (±10°) CLUSTERING (TRAIN)
# ==============================================================================
n_drop_fcm = int(dropped.sum())
n_ok_fcm = int((in_target & (target_prob >= best_fcm_threshold)).sum())
print(f"FCM optimized (train): noise = {n_drop_fcm}/{len(df)}, OK cluster = {n_ok_fcm}")

EPS_GRID = np.round(np.arange(0.10, 2.0 + 1e-9, 0.02), 2)
MS_GRID = range(3, 16)

best_dbm = None
for eps in EPS_GRID:
    for ms in MS_GRID:
        lab = DBSCAN(eps=eps, min_samples=ms).fit_predict(X_scaled)
        ids = np.unique(lab[lab != -1])
        if len(ids) < 2:
            continue
        d = np.linalg.norm(np.array([raw_angles[lab == i].mean(0) for i in ids]) - TARGET_ANGLES, axis=1)
        k = int(d.argmin())
        n_ok = int((lab == ids[k]).sum())
        n_noise = int((lab == -1).sum())
        key = (abs(n_noise - n_drop_fcm), abs(n_ok - n_ok_fcm), d[k])
        if best_dbm is None or key < best_dbm[0]:
            best_dbm = (key, eps, ms, lab, ids[k], n_ok, n_noise)

_, best_dbm_eps, best_dbm_ms, dbm_labels, dbm_ok_id, dbm_n_ok, dbm_n_noise = best_dbm
print(f"DBSCAN matched retention -> eps: {best_dbm_eps} | min_samples: {best_dbm_ms} | "
      f"noise = {dbm_n_noise} (FCM: {n_drop_fcm}) | OK cluster = {dbm_n_ok} (FCM: {n_ok_fcm}) | "
      f"{'EQUAL noise' if dbm_n_noise == n_drop_fcm else 'noise NOT equal - refine EPS_GRID'}")

df['dbscan_matched retention_cluster'] = dbm_labels
print(df['dbscan_matched retention_cluster'].value_counts().sort_index())

# ---- Rule-based (±10° tolerance) pseudo-cluster assignment ----
# Cluster 0 = all 4 joints within ±TOL of TARGET_ANGLES (Correct posture)
# Clusters 1..4 = grouped by primary offending joint (1: Elbow, 2: Shoulder, 3: Hip, 4: Knee)
def assign_rule_clusters(angles_arr, tol=TOL):
    ex = np.maximum(np.abs(angles_arr - TARGET_ANGLES) - tol, 0.0)
    is_ok = ex.max(axis=1) == 0
    primary_err_joint = np.argmax(ex, axis=1) + 1
    return np.where(is_ok, 0, primary_err_joint)

df['rule_cluster'] = assign_rule_clusters(raw_angles, tol=TOL)
print(f"\n{RULE_NAME} train cluster counts (0=OK, 1=Elbow, 2=Shoulder, 3=Hip, 4=Knee):")
print(df['rule_cluster'].value_counts().sort_index())


# %%
# ==============================================================================
# 12. SAVE TRAIN CLUSTERING RESULTS (ALL ALGORITHMS + RULE-BASED)
# ==============================================================================
label_cols = [
    'kmeans_cluster', 'dbscan_cluster', 'hierarchical_cluster', 'gmm_cluster',
    'dbscan_matched retention_cluster', 'fcm_cluster', 'fcm_opt_cluster', 'fcm_grid_cluster',
    'rule_cluster'
]
prob_cols = sorted(c for c in df.columns if c.startswith('Prob_Cluster_'))
id_cols = [c for c in ['filename', 'head_side'] if c in df.columns]

train_results = df[id_cols + feature_columns + label_cols + prob_cols]
train_results.to_csv('train_clustering_results.csv', index=False)
print(f"Saved {train_results.shape[0]} train samples x {train_results.shape[1]} columns to 'train_clustering_results.csv'")
display(train_results.head())


# %%
# ==============================================================================
# 13. TEST SET CLUSTER ASSIGNMENTS (ALL ALGORITHMS + RULE-BASED)
# ==============================================================================
test = pd.read_csv('test.csv')
V = test[feature_columns].to_numpy(float)
Xt = scaler.transform(test[feature_columns])  # scaler fitted on TRAIN only

test_clusters = test[['filename']].copy() if 'filename' in test.columns else pd.DataFrame(index=test.index)

# K-Means & GMM
test_clusters['kmeans_cluster'] = final_kmeans.predict(Xt)
test_clusters['gmm_cluster'] = final_gmm.predict(Xt)

# Hierarchical (nearest TRAIN cluster centroid)
def centroid_predict(train_labels, Xtest):
    ids = np.unique(train_labels)
    cents = np.array([X_scaled[train_labels == i].mean(0) for i in ids])
    return ids[np.linalg.norm(Xtest[:, None] - cents[None], axis=2).argmin(1)]

test_clusters['hierarchical_cluster'] = centroid_predict(df['hierarchical_cluster'].to_numpy(), Xt)

# DBSCAN (nearest TRAIN core point within eps, otherwise noise = -1)
def dbscan_predict(eps, min_samples, Xtest):
    db = DBSCAN(eps=eps, min_samples=min_samples).fit(X_scaled)
    core, core_lab = X_scaled[db.core_sample_indices_], db.labels_[db.core_sample_indices_]
    dist = np.linalg.norm(Xtest[:, None] - core[None], axis=2)
    return np.where(dist.min(1) <= eps, core_lab[dist.argmin(1)], -1)

test_clusters['dbscan_cluster'] = dbscan_predict(best_db_eps, best_db_min_samples, Xt)
test_clusters['dbscan_matched retention_cluster'] = dbscan_predict(best_dbm_eps, best_dbm_ms, Xt)

# FCM & Thresholded FCM
test_mem = fcm_membership(Xt, final_fcm_centers, best_fcm_m)
test_fcm_label = test_mem.argmax(1)
test_fcm_prob = test_mem.max(1) * 100

test_clusters['fcm_cluster'] = test_fcm_label
test_clusters['fcm_opt_cluster'] = np.where(
    (test_fcm_label == fcm_target) & (test_fcm_prob < best_fcm_threshold), FCM_DISCARD_ID, test_fcm_label
)
test_clusters['fcm_grid_cluster'] = np.where(
    (test_fcm_label == fcm_target) & (test_fcm_prob < best_grid_threshold), FCM_DISCARD_ID, test_fcm_label
)

# Rule-based (±10° tolerance)
test_clusters['rule_cluster'] = assign_rule_clusters(V, tol=TOL)

# Save & display
test_clusters = pd.concat([test_clusters, test[feature_columns]], axis=1)
test_clusters.to_csv('test_cluster_assignments.csv', index=False)
display(test_clusters.head())

for col in label_cols:
    print(f"{col:24s}:", test_clusters[col].value_counts().sort_index().to_dict())


# %%
# ==============================================================================
# 14. IDENTIFY CORRECT (OK) POSTURE CLUSTER FOR EACH METHOD (INCLUDING RULE-BASED)
# ==============================================================================
methods = {
    'KMeans':           ('kmeans_cluster',         []),
    'DBSCAN':           ('dbscan_cluster',         [-1]),
    'Hierarchical':     ('hierarchical_cluster',   []),
    'GMM':              ('gmm_cluster',            []),
    'DBSCAN (matched retention)': ('dbscan_matched retention_cluster', [-1]),
    'FCM':              ('fcm_cluster',            []),
    'FCM + NSGA-II':    ('fcm_opt_cluster',        [FCM_DISCARD_ID]),
    'fcm + grid search':       ('fcm_grid_cluster',       [FCM_DISCARD_ID]),
    RULE_NAME:          ('rule_cluster',           []),
}

correct_cluster = {}
rows = []
for name, (col, excluded) in methods.items():
    sub = df[~df[col].isin(excluded)]
    if name == RULE_NAME and (sub[col] == 0).any():
        cid = 0
        means = sub.groupby(col)[feature_columns].mean()
        mu = means.loc[cid].values
        dist_val = float(np.linalg.norm(mu - TARGET_ANGLES))
    else:
        means = sub.groupby(col)[feature_columns].mean()
        dist = np.linalg.norm(means.values - TARGET_ANGLES, axis=1)
        cid = int(means.index[dist.argmin()])
        mu = means.loc[cid].values
        dist_val = float(dist.min())

    correct_cluster[name] = int(cid)
    rows.append({
        'Method': name,
        'Correct cluster': int(cid),
        'n': int((sub[col] == cid).sum()),
        **{f.split('_')[0].title(): round(v, 1) for f, v in zip(feature_columns, mu)},
        'Total error (deg)': round(dist_val, 2)
    })

correct_table = pd.DataFrame(rows).set_index('Method')
display(correct_table)
correct_table.to_csv('table_correct_clusters.csv')
print("Correct cluster IDs:", correct_cluster)


# %%
# ==============================================================================
# 15. TWO BIGGEST ERRORS OF EACH CLUSTER VS. CORRECT CLUSTER RANGE
# ==============================================================================
rows = []
for name, (col, excluded) in methods.items():
    cid = correct_cluster[name]
    ok = df[df[col] == cid][feature_columns]

    if name == RULE_NAME:
        mu = TARGET_ANGLES.astype(float)
        sd = np.full(4, TOL / K_STD)  # so lo = TARGET - TOL, hi = TARGET + TOL
    else:
        mu = ok.mean().values
        sd = np.maximum(ok.std(ddof=1).fillna(0.0).values, 1e-6)

    lo, hi = mu - K_STD * sd, mu + K_STD * sd

    for c in sorted(df[col].unique()):
        if c == cid or c in excluded:
            continue
        sub = df[df[col] == c]
        v = sub[feature_columns].mean().values
        out_deg = np.where(v < lo, lo - v, np.where(v > hi, v - hi, 0.0))
        out_sd = out_deg / sd
        top = [j for j in np.argsort(-out_sd)[:MAX_ERRORS] if out_sd[j] > 0]

        desc = ', '.join(
            f"{short[j]} {'high' if v[j] > hi[j] else 'low'} ({out_deg[j]:.1f}° out, {out_sd[j]:.1f} SD)"
            for j in top
        ) or 'none (inside range)'

        rows.append({
            'Method': name,
            'Cluster': int(c),
            'n': len(sub),
            **{s: round(x, 1) for s, x in zip(short, v)},
            'Biggest errors': desc
        })

cluster_errors = pd.DataFrame(rows)
display(cluster_errors)
cluster_errors.to_csv('cluster_error_angles.csv', index=False)


# %%
# ==============================================================================
# 16. EVALUATION VS. EXPERT LABELS (TABLE 1: OWN SAMPLES & TABLE 2: COMMON SAMPLES)
# ==============================================================================
expert = pd.read_csv('expert_labeling.csv').set_index('filename').loc[test['filename']].reset_index()
E = expert[feature_columns].notna().astype(int).values

def predict_rule(tol=TOL):
    ex = np.maximum(np.abs(V - TARGET_ANGLES) - tol, 0.0)
    P = np.zeros((len(test), 4), int)
    for i in range(len(test)):
        if ex[i].max() > 0:
            P[i, [j for j in np.argsort(-ex[i])[:MAX_ERRORS] if ex[i, j] > 0]] = 1
    return P, np.ones(len(test), bool)

def predict_errors(name):
    if name == RULE_NAME:
        return predict_rule(tol=TOL)
    col, excluded = methods[name]
    labels = test_clusters[col].to_numpy()
    ok = df[df[col] == correct_cluster[name]][feature_columns]
    mu, sd = ok.mean().values, np.maximum(ok.std(ddof=1).fillna(0.0).values, 1e-6)
    lo, hi = mu - K_STD * sd, mu + K_STD * sd
    P = np.zeros((len(test), 4), int)
    keep = ~np.isin(labels, excluded)
    for i, c in enumerate(labels):
        if not keep[i] or c == correct_cluster[name]:
            continue
        ex = np.where(V[i] < lo, (lo - V[i]) / sd, np.where(V[i] > hi, (V[i] - hi) / sd, 0.0))
        ch = [j for j in np.argsort(-ex)[:MAX_ERRORS] if ex[j] > 0] or [int(np.argmax(np.abs(V[i] - mu) / sd))]
        P[i, ch] = 1
    return P, keep

results = {name: predict_errors(name) for name in methods}

def metrics(P, keep):
    p, e = P[keep], E[keep]
    has_err = e.sum(1) > 0
    ok_e, ok_p = ~has_err, p.sum(1) == 0
    tp, fn = int((ok_e & ok_p).sum()),  int((ok_e & ~ok_p).sum())
    fp, tn = int((~ok_e & ok_p).sum()), int((~ok_e & ~ok_p).sum())
    return {
        'Samples used': int(keep.sum()),
        'Excluded (noise)': int((~keep).sum()),
        'Overall agreement': (p == e).mean(),
        'Agreement on non-correct samples': (p[has_err] == e[has_err]).mean(),
        'Agreement on correct samples': (p[~has_err] == e[~has_err]).mean(),
        'TP': tp, 'FN': fn, 'FP': fp, 'TN': tn,
        'Accuracy': (tp + tn) / max(len(ok_e), 1),
        'Precision': tp / max(tp + fp, 1),
        'Recall': tp / max(tp + fn, 1),
        'Specificity': tn / max(tn + fp, 1),
        'F1': 2 * tp / max(2 * tp + fp + fn, 1),
        'Kappa': cohen_kappa_score(ok_e, ok_p)
    }

# Table 1: Each method on its own kept samples
table1 = pd.DataFrame({n: metrics(P, keep) for n, (P, keep) in results.items()}).T
table1 = table1.astype({c: int for c in ['Samples used', 'Excluded (noise)', 'TP', 'FN', 'FP', 'TN']})

# Table 2: Common non-noise samples across all methods
common_keep = np.ones(len(test), bool)
for P, keep in results.values():
    common_keep &= keep
print(f"Test samples: {len(test)} | kept by every method: {int(common_keep.sum())} | removed: {int((~common_keep).sum())}")

table2 = pd.DataFrame({n: metrics(P, common_keep) for n, (P, keep) in results.items()}).T
table2 = table2.astype({c: int for c in ['Samples used', 'Excluded (noise)', 'TP', 'FN', 'FP', 'TN']})

print("\nTABLE 1 - each method on its own kept samples (noise of DBSCAN / FCM-threshold excluded)")
display(table1.round(3))
print("\nTABLE 2 - all methods on the common non-noise samples")
display(table2.round(3))

table1.round(4).to_csv('eval_table1_own_samples.csv')
table2.round(4).to_csv('eval_table2_common_samples.csv')


# %%
# ==============================================================================
# 17. PER-CLUSTER DEVIATIONS & OK-CLUSTER LEADERBOARD (INCLUDING RULE-BASED)
# ==============================================================================
# Table 1 style: every cluster of every algorithm (including Rule-based)
rows = []
for name, (col, excluded) in methods.items():
    for c in sorted(df[col].unique()):
        sub = df[df[col] == c]
        dev = sub[feature_columns].mean().values - TARGET_ANGLES
        status = 'Outlier' if c in excluded else ('Target' if c == correct_cluster[name] else 'Problem')
        rows.append({
            'Method': name,
            'Cluster ID': int(c),
            'Sample Count (n)': len(sub),
            'Status': status,
            **{f'{s} Deviation': f"{d:+.1f}°" for s, d in zip(short, dev)}
        })
table_dev = pd.DataFrame(rows)
display(table_dev)
table_dev.to_csv('table_cluster_deviations_all_methods.csv', index=False)

# Table 2 style: OK-cluster leaderboard (including Rule-based)
rows = []
for name, (col, excluded) in methods.items():
    ok = df[df[col] == correct_cluster[name]][feature_columns]
    mu, sd = ok.mean().values, ok.std(ddof=1).fillna(0.0).values
    rows.append({
        'Model / Stage': name,
        'Sample Count (n)': len(ok),
        'Total Cumulative Error (°)': round(float(np.linalg.norm(mu - TARGET_ANGLES)), 2),
        **{f'{s} (Mean ± SD)': f"{m:.1f}° (±{v:.1f})" for s, m, v in zip(short, mu, sd)}
    })
leaderboard = pd.DataFrame(rows).sort_values('Total Cumulative Error (°)').reset_index(drop=True)
display(leaderboard)
leaderboard.to_csv('table_error_leaderboard.csv', index=False)


# %%
# ==============================================================================
# 18. 3D PCA PROJECTIONS OF CLUSTERS (ALL 9 METHODS + INDIVIDUAL PLOTS)
# ==============================================================================
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
var = pca.explained_variance_ratio_ * 100

TARGET_COLOR = '#009E73'  # Emerald/teal
NOISE_COLOR  = '#B0B0B0'  # Light neutral gray
error_palette = [
    '#E63946', '#FF8C00', '#4361EE', '#9B5DE5', '#F1C40F',
    '#00B4D8', '#D81159', '#8D6E63', '#1D3557', '#AACC00',
]

def draw_pca(ax, name, col, excluded):
    labels = df[col].to_numpy()
    err_idx = 0
    for c in sorted(np.unique(labels)):
        m = labels == c
        if c in excluded:
            ax.scatter(
                X_pca[m, 0], X_pca[m, 1], X_pca[m, 2],
                c=NOISE_COLOR, alpha=0.35, s=14,
                label=f'Noise/discarded (n={m.sum()})'
            )
        elif c == correct_cluster[name]:
            ax.scatter(
                X_pca[m, 0], X_pca[m, 1], X_pca[m, 2],
                c=TARGET_COLOR, alpha=0.85, s=24, edgecolors='k', linewidth=0.3,
                label=f'Target C{c} (n={m.sum()})'
            )
        else:
            color = error_palette[err_idx % len(error_palette)]
            err_idx += 1
            err_lbl = f'Error C{c} ({short[c-1]}) (n={m.sum()})' if name == RULE_NAME and 1 <= c <= 4 else f'Error C{c} (n={m.sum()})'
            ax.scatter(
                X_pca[m, 0], X_pca[m, 1], X_pca[m, 2],
                c=color, alpha=0.65, s=18,
                label=err_lbl
            )

    ax.set_title(name, fontweight='bold')
    ax.set_xlabel(f'PC1 ({var[0]:.1f}%)')
    ax.set_ylabel(f'PC2 ({var[1]:.1f}%)')
    ax.set_zlabel(f'PC3 ({var[2]:.1f}%)')
    ax.view_init(elev=20, azim=55)
    ax.legend(fontsize=7, loc='upper left', bbox_to_anchor=(0.0, 0.95))

# Combined 3x3 grid for all 9 methods (including Rule-based)
fig = plt.figure(figsize=(22, 18), dpi=200)
for i, (name, (col, excluded)) in enumerate(methods.items(), 1):
    ax = fig.add_subplot(3, 3, i, projection='3d')
    draw_pca(ax, name, col, excluded)
plt.suptitle('3D PCA projection of posture clusters ', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('pca3d_all_algorithms.png', dpi=300, bbox_inches='tight')
plt.show()

# Individual high-res 3D PCA figure per method
for name, (col, excluded) in methods.items():
    fig = plt.figure(figsize=(9, 7), dpi=200)
    ax = fig.add_subplot(111, projection='3d')
    draw_pca(ax, name, col, excluded)
    plt.tight_layout()
    safe_name = (name.replace(' ', '_').replace('+', 'plus')
                     .replace('(', '').replace(')', '')
                     .replace('±', 'pm').replace('°', 'deg'))
    plt.savefig(f"pca3d_{safe_name}.png", dpi=300, bbox_inches='tight')
    plt.close(fig)


# %%
# ==============================================================================
# 19. KDE DISTRIBUTIONS OF THE OK-CLUSTER ANGLES (ALL 9 METHODS)
# ==============================================================================
sns.set_theme(style='whitegrid')

ok_data = {name: df[df[col] == correct_cluster[name]] for name, (col, _) in methods.items()}
colors = sns.color_palette('tab10', len(ok_data))

# Diagram 1: Per joint, all algorithms + Rule-based on the same axes
fig, axes = plt.subplots(2, 2, figsize=(16, 11), dpi=200)
for k, (ax, feat) in enumerate(zip(axes.ravel(), feature_columns)):
    for (name, sub), colr in zip(ok_data.items(), colors):
        if len(sub) > 1 and sub[feat].std() > 1e-6:
            sns.kdeplot(sub[feat], ax=ax, color=colr, lw=2, label=f"{name} (n={len(sub)})", warn_singular=False)
    ax.axvline(TARGET_ANGLES[k], color='k', ls='--', lw=1.8, label=f'Ideal ({TARGET_ANGLES[k]}°)')
    if k in (0, 1):
        ax.axvspan(TARGET_ANGLES[k] - TOL, TARGET_ANGLES[k] + TOL, color='gray', alpha=0.12, label=f'Rule ±{TOL:.0f}° band')
    else:
        ax.axvspan(TARGET_ANGLES[k] - TOL, TARGET_ANGLES[k] + TOL, color='gray', alpha=0.12)
    ax.set_title(feat, fontweight='bold')
    ax.set_xlabel('Angle (degrees)')
    ax.set_ylabel('Density')
axes[0, 0].legend(fontsize=8)
plt.suptitle('OK-cluster angle distributions: all algorithms & rule-based tolerance', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig('kde_ok_cluster_all_models_per_angle.png', dpi=300, bbox_inches='tight')
plt.show()

# Diagram 2: 3x3 grid (one panel per algorithm, all 4 joints together)
fig, axes = plt.subplots(3, 3, figsize=(22, 15), dpi=200)
for ax, (name, sub) in zip(axes.ravel(), ok_data.items()):
    for feat in feature_columns:
        if len(sub) > 1 and sub[feat].std() > 1e-6:
            sns.kdeplot(sub[feat], ax=ax, fill=True, alpha=0.2, lw=2,
                        label=feat.split('_')[0].title(), warn_singular=False)
    ax.set_title(f"{name}\n(correct form cluster {correct_cluster[name]}, n={len(sub)})",
                 fontweight='bold', fontsize=11)
    ax.set_xlabel('Angle (degrees)')
    ax.set_ylabel('Density')
axes[0, 0].legend(fontsize=8)
plt.suptitle('Correct form cluster angle distributions per method (including Rule-based)', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig('kde_ok_cluster_per_model.png', dpi=300, bbox_inches='tight')
plt.show()


# %%
# ==============================================================================
# 20. STYLED PAPER LEADERBOARD TABLE (BOLD BEST VALUES + CSV & EXCEL EXPORT)
# ==============================================================================
paper_names = {
    'KMeans':           'K-Means',
    'DBSCAN':           'DBSCAN (Tuned with NSGA-II)',
    'Hierarchical':     'Hierarchical',
    'GMM':              'GMM',
    'DBSCAN (matched retention)': 'DBSCAN (same retention as FCM-opt)',
    'FCM':              'FCM (pre-optimization)',
    'FCM + NSGA-II':    'FCM (Optimized with NSGA-II)',
    'fcm + grid search':       'FCM (Grid-search threshold)',
    RULE_NAME:          f'Rule-based tolerance (±{TOL:.0f}°)',
}

num = []
for name, (col, _) in methods.items():
    ok = df[df[col] == correct_cluster[name]][feature_columns]
    mu, sd = ok.mean().values, ok.std(ddof=1).fillna(0.0).values
    num.append({
        'Model / Stage': paper_names[name],
        'n': len(ok),
        'Error': float(np.linalg.norm(mu - TARGET_ANGLES)),
        **{f'{s}_mu': m for s, m in zip(short, mu)},
        **{f'{s}_sd': v for s, v in zip(short, sd)}
    })
num = pd.DataFrame(num).sort_values('Error').reset_index(drop=True)

best_col = {'n': num['n'].idxmax(), 'Error': num['Error'].idxmin()}
for k, s in enumerate(short):
    best_col[f'{s}_mu'] = (num[f'{s}_mu'] - TARGET_ANGLES[k]).abs().idxmin()

paper = pd.DataFrame({
    'Model / Stage': num['Model / Stage'],
    'Sample Count (n)': num['n'],
    'Total Cumulative Error (°)': num['Error'].round(2)
})
for s in short:
    paper[f'{s} (Mean ± SD)'] = [f"{m:.1f}° (±{v:.1f})" for m, v in zip(num[f'{s}_mu'], num[f'{s}_sd'])]

def bold_best(data):
    styles = pd.DataFrame('', index=data.index, columns=data.columns)
    styles.loc[best_col['n'], 'Sample Count (n)'] = 'font-weight: bold'
    styles.loc[best_col['Error'], 'Total Cumulative Error (°)'] = 'font-weight: bold'
    for s in short:
        styles.loc[best_col[f'{s}_mu'], f'{s} (Mean ± SD)'] = 'font-weight: bold'
    return styles

display(paper)
paper.to_csv('paper_table_ok_cluster_leaderboard.csv', index=False)

try:
    styled = paper.style.apply(bold_best, axis=None).hide(axis='index')
    styled.to_excel('paper_table_ok_cluster_leaderboard.xlsx', engine='openpyxl', index=False)
except Exception as e:
    print(f"Note: Could not save styled .xlsx ({e}); CSV saved to 'paper_table_ok_cluster_leaderboard.csv'.")


# %%
# ==============================================================================
# 21. MULTI-SEED EVALUATION (ALL 9 METHODS x 5 SEEDS -> MEAN ± SD & PAIRED T-TESTS)
# ==============================================================================
order = [
    'KMeans', 'DBSCAN', 'Hierarchical', 'GMM', 'DBSCAN (matched retention)',
    'FCM', 'FCM + NSGA-II', 'fcm + grid search', RULE_NAME
]

hier_train = df['hierarchical_cluster'].to_numpy()
hier_test = centroid_predict(hier_train, Xt)
rule_train = df['rule_cluster'].to_numpy()
rule_test = test_clusters['rule_cluster'].to_numpy()

db_cands = []
for eps in EPS_GRID:
    for ms in MS_GRID:
        lab = DBSCAN(eps=eps, min_samples=ms).fit_predict(X_scaled)
        ids = np.unique(lab[lab != -1])
        if len(ids) < 2:
            continue
        d = np.linalg.norm(np.array([raw_angles[lab == i].mean(0) for i in ids]) - TARGET_ANGLES, axis=1)
        k = int(d.argmin())
        db_cands.append((eps, ms, lab, int((lab == ids[k]).sum()), int((lab == -1).sum()), float(d[k])))

def knee(F):
    Fn = (F - F.min(0)) / (F.max(0) - F.min(0) + 1e-12)
    return int(np.linalg.norm(Fn, axis=1).argmin())

def ok_cluster(lab, excluded):
    ids = [i for i in np.unique(lab) if i not in excluded]
    d = [np.linalg.norm(raw_angles[lab == i].mean(0) - TARGET_ANGLES) for i in ids]
    return int(ids[int(np.argmin(d))])

def predict_errors_gen(train_lab, test_lab, ok_id, excluded):
    ok = df[train_lab == ok_id][feature_columns]
    mu, sd = ok.mean().values, np.maximum(ok.std(ddof=1).fillna(0.0).values, 1e-6)
    lo, hi = mu - K_STD * sd, mu + K_STD * sd
    keep = ~np.isin(test_lab, excluded)
    P = np.zeros((len(test), 4), int)
    for i, c in enumerate(test_lab):
        if not keep[i] or c == ok_id:
            continue
        ex = np.where(V[i] < lo, (lo - V[i]) / sd, np.where(V[i] > hi, (V[i] - hi) / sd, 0.0))
        ch = [j for j in np.argsort(-ex)[:MAX_ERRORS] if ex[j] > 0] or [int(np.argmax(np.abs(V[i] - mu) / sd))]
        P[i, ch] = 1
    return P, keep

STORE = {}

def run_seed(seed):
    M = {}  # name: (train labels, test labels, excluded ids)

    # K-Means & GMM
    km = KMeans(n_clusters=best_kmeans_k, random_state=seed, n_init=20).fit(X_scaled)
    M['KMeans'] = (km.labels_, km.predict(Xt), [])
    gm = GaussianMixture(n_components=best_gmm_k, random_state=seed, n_init=5, reg_covar=1e-6).fit(X_scaled)
    M['GMM'] = (gm.predict(X_scaled), gm.predict(Xt), [])

    # Hierarchical (deterministic)
    M['Hierarchical'] = (hier_train, hier_test, [])

    # DBSCAN tuned with NSGA-II
    r = minimize(
        DBSCANProblem(X_scaled),
        NSGA2(pop_size=40, sampling=FloatRandomSampling(), crossover=SBX(prob=0.9, eta=15),
              mutation=PM(eta=20), eliminate_duplicates=True),
        ('n_gen', 50), seed=seed, verbose=False
    )
    Fd, Xd = np.atleast_2d(r.F), np.atleast_2d(r.X)
    j = knee(Fd)
    e_, m_ = float(Xd[j][0]), int(round(Xd[j][1]))
    M['DBSCAN'] = (DBSCAN(eps=e_, min_samples=m_).fit_predict(X_scaled), dbscan_predict(e_, m_, Xt), [-1])

    # FCM & Thresholds
    centers, *_ = fuzz.cluster.cmeans(
        fcm_data, c=best_fcm_c, m=best_fcm_m, error=0.005, maxiter=1000, init=None, seed=seed
    )
    tr_mem, te_mem = fcm_membership(X_scaled, centers, best_fcm_m), fcm_membership(Xt, centers, best_fcm_m)
    tr_lab, te_lab = tr_mem.argmax(1), te_mem.argmax(1)
    means = np.array([
        raw_angles[tr_lab == k].mean(0) if (tr_lab == k).any() else np.full(4, np.inf)
        for k in range(best_fcm_c)
    ])
    tgt = int(np.linalg.norm(means - TARGET_ANGLES, axis=1).argmin())
    in_t, tp_, te_prob = (tr_lab == tgt), tr_mem[:, tgt] * 100, te_mem[:, tgt] * 100
    DISC = best_fcm_c
    M['FCM'] = (tr_lab, te_lab, [])

    def thr_obj(t):
        m = in_t & (tp_ >= t)
        if not m.any():
            return [1.0, 10000.0]
        return [1.0 - m.sum() / len(df), np.mean((raw_angles[m].mean(0) - TARGET_ANGLES) ** 2)]

    class TP(ElementwiseProblem):
        def __init__(s):
            super().__init__(n_var=1, n_obj=2, xl=np.array([50.0]), xu=np.array([99.9]))
        def _evaluate(s, x, out, *a, **k):
            out["F"] = thr_obj(x[0])

    r = minimize(
        TP(),
        NSGA2(pop_size=40, n_offsprings=10, sampling=FloatRandomSampling(),
              crossover=SBX(prob=0.9, eta=15), mutation=PM(eta=20), eliminate_duplicates=True),
        ('n_gen', 50), seed=seed, verbose=False
    )
    thr_n = float(np.atleast_2d(r.X)[knee(np.atleast_2d(r.F))][0])

    Fg = np.array([thr_obj(t) for t in GRID])
    pm_g = pareto_mask(Fg)
    thr_g = float(GRID[pm_g][knee(Fg[pm_g])])

    for name, thr in [('FCM + NSGA-II', thr_n), ('fcm + grid search', thr_g)]:
        M[name] = (
            np.where(in_t & (tp_ < thr), DISC, tr_lab),
            np.where((te_lab == tgt) & (te_prob < thr), DISC, te_lab),
            [DISC]
        )

    # DBSCAN matched retention to FCM + NSGA-II retention
    n_drop = int((M['FCM + NSGA-II'][0] == DISC).sum())
    n_ok = int((M['FCM + NSGA-II'][0] == tgt).sum())
    eps, ms, db_lab, _, _, _ = min(db_cands, key=lambda c: (abs(c[4] - n_drop), abs(c[3] - n_ok), c[5]))
    M['DBSCAN (matched retention)'] = (db_lab, dbscan_predict(eps, ms, Xt), [-1])

    # Rule-based (±10° tolerance)
    M[RULE_NAME] = (rule_train, rule_test, [])

    # Evaluate on test set
    res = {}
    for n, (tr, te, ex) in M.items():
        if n == RULE_NAME:
            res[n] = predict_rule(tol=TOL)
        else:
            res[n] = predict_errors_gen(tr, te, ok_cluster(tr, ex), ex)

    common = np.logical_and.reduce([k for _, k in res.values()])

    rows = []
    for name, (P, keep) in res.items():
        row = {'Method': name, 'Seed': seed}
        if name == 'FCM + NSGA-II':
            row['Threshold'] = thr_n
        if name == 'fcm + grid search':
            row['Threshold'] = thr_g

        tr, _, ex = M[name]
        ok_id = 0 if (name == RULE_NAME and (tr == 0).any()) else ok_cluster(tr, ex)
        ok = raw_angles[tr == ok_id]
        mu = ok.mean(0)
        sd_vals = ok.std(0, ddof=1) if len(ok) > 1 else np.zeros(4)

        row.update({
            'n (OK cluster)': len(ok),
            'Train noise': int(np.isin(tr, ex).sum()),
            'Total error (deg)': float(np.linalg.norm(mu - TARGET_ANGLES))
        })
        row.update({f'{s}_mu': v for s, v in zip(short, mu)})
        row.update({f'{s}_sd': v for s, v in zip(short, sd_vals)})
        row.update(metrics(P, keep))
        row.update({k + ' [T2]': v for k, v in metrics(P, common).items()})
        rows.append(row)

    STORE[seed] = (M, res)
    return rows

rows = []
for s in RANDOM_SEEDS:
    rows += run_seed(s)
    print('Completed seed:', s)

runs = pd.DataFrame(rows)
runs.round(4).to_csv('multiseed_all_methods_per_seed.csv', index=False)

cols = [c for c in runs.columns if c not in ('Method', 'Seed')]
mean_ = runs.groupby('Method')[cols].mean().loc[order]
sd_ = runs.groupby('Method')[cols].std(ddof=1).fillna(0.0).loc[order]

def fmt(c, d=3):
    return mean_[c].map(lambda v: f'{v:.{d}f}') + ' ± ' + sd_[c].map(lambda v: f'{v:.{d}f}')

# Averaged Leaderboard (saved to CSV)
lb = pd.DataFrame({
    'Sample Count (n)': fmt('n (OK cluster)', 1),
    'Total Cumulative Error (°)': fmt('Total error (deg)', 2)
})
for s in short:
    lb[f'{s} (Mean ± SD)'] = (
        mean_[f'{s}_mu'].map('{:.1f}°'.format) + ' (±' + mean_[f'{s}_sd'].map('{:.1f}'.format) + ')'
    )
display(lb)
lb.to_csv('avg_leaderboard.csv')

# Averaged Table 1 & Table 2 Test Metrics (saved to CSV)
mcols = [
    'Samples used', 'Excluded (noise)', 'Overall agreement', 'Agreement on non-correct samples',
    'Agreement on correct samples', 'TP', 'FN', 'FP', 'TN', 'Accuracy', 'Precision', 'Recall',
    'Specificity', 'F1', 'Kappa'
]
t1 = pd.DataFrame({c: fmt(c) for c in mcols})
t2 = pd.DataFrame({c: fmt(c + ' [T2]') for c in mcols})
print('TABLE 1 - own kept samples, mean ± SD over 5 seeds')
display(t1)
t1.to_csv('avg_table1_own_samples.csv')

print('TABLE 2 - common non-noise samples, mean ± SD over 5 seeds')
display(t2)
t2.to_csv('avg_table2_common_samples.csv')

# Paired comparisons: FCM + NSGA-II vs. every other method (saved to CSV)
piv = runs.pivot(index='Seed', columns='Method', values='Total error (deg)')
ttest_rows = []
print("\nPaired comparison vs. FCM + NSGA-II (Total Cumulative Error):")
for other in ['KMeans', 'DBSCAN', 'Hierarchical', 'GMM', 'DBSCAN (matched retention)', 'FCM', 'fcm + grid search', RULE_NAME]:
    d = piv[other] - piv['FCM + NSGA-II']
    sdv = float(d.std(ddof=1))
    p_val = float(stats.ttest_rel(piv[other], piv['FCM + NSGA-II']).pvalue) if sdv > 1e-12 else float('nan')
    wins = int((d > 0).sum())
    print(f'{other:18s} gap {d.mean():+.3f} ± {sdv:.3f} deg | FCM+NSGA-II better in {wins}/5 seeds | paired t p={p_val:.3f}')
    ttest_rows.append({
        'Comparison Method': other,
        'Error Gap Mean (deg)': round(float(d.mean()), 4),
        'Error Gap SD (deg)': round(sdv, 4),
        'FCM+NSGA-II Wins (out of 5)': f"{wins}/5",
        'Paired t-test p-value': round(p_val, 4) if not np.isnan(p_val) else np.nan
    })

paired_ttest_df = pd.DataFrame(ttest_rows)
paired_ttest_df.to_csv('table_paired_ttest_vs_fcm_nsga2.csv', index=False)

# Membership thresholds per seed & summary (saved to CSV)
thr = runs[runs.Method.isin(['FCM + NSGA-II', 'fcm + grid search'])].pivot(index='Seed', columns='Method', values='Threshold')
thr_summary = thr.agg(['mean', 'std']).round(2)
print('\nMembership thresholds per seed (%):')
display(thr.round(2))
print('mean ± SD:')
display(thr_summary)

thr.round(4).to_csv('table_fcm_thresholds_per_seed.csv')
thr_summary.to_csv('table_fcm_thresholds_summary.csv')


# %%
# ==============================================================================
# 22. PER-CLUSTER & ROLE-LEVEL TEST AGREEMENT SUMMARY (SAVED TO CSV)
# ==============================================================================
def cluster_table(seed, name='FCM + NSGA-II'):
    (M, res) = STORE[seed]
    tr, te, ex = M[name]
    P, keep = res[name]
    ok_id = 0 if (name == RULE_NAME and (tr == 0).any()) else ok_cluster(tr, ex)
    agree = (P == E).mean(1)
    rows = []
    for c in np.unique(te):
        m = te == c
        disc = c in ex
        rows.append({
            'Cluster': int(c),
            'Status': 'Discarded' if disc else ('Target' if c == ok_id else 'Problem'),
            'Test n': int(m.sum()),
            'Expert-correct (n)': int((E[m].sum(1) == 0).sum()),
            'Avg joint agreement (%)': np.nan if disc else round(100 * agree[m].mean(), 2)
        })
    return pd.DataFrame(rows)

ct_fcm42 = cluster_table(42, 'FCM + NSGA-II')
print('Per-cluster test table for FCM + NSGA-II (seed 42):')
display(ct_fcm42)
ct_fcm42.to_csv('table_per_cluster_test_fcm_nsga2_seed42.csv', index=False)

ct_rule42 = cluster_table(42, RULE_NAME)
print(f'\nPer-cluster test table for {RULE_NAME} (seed 42):')
display(ct_rule42)
ct_rule42.to_csv('table_per_cluster_test_rule_seed42.csv', index=False)

role_rows = []
for s in RANDOM_SEEDS:
    t = cluster_table(s, 'FCM + NSGA-II')
    for role, g in t.groupby('Status'):
        n = g['Test n'].sum()
        agr = np.nansum(g['Avg joint agreement (%)'] * g['Test n']) / n if role != 'Discarded' else np.nan
        role_rows.append({'Seed': s, 'Role': role, 'n': n, 'Agreement (%)': agr})
rr = pd.DataFrame(role_rows)
role_summary = rr.groupby('Role')[['n', 'Agreement (%)']].agg(['mean', 'std']).round(2)
print('\nFCM + NSGA-II by role, mean ± SD over 5 seeds:')
display(role_summary)
role_summary.to_csv('table_role_summary_fcm_nsga2_5seeds.csv')


# %%
# ==============================================================================
# 23. BOOTSTRAP 95% CONFIDENCE INTERVALS (ALL 9 METHODS INCLUDING RULE-BASED)
# ==============================================================================
def bin_arrays(name, seed):
    (M, res) = STORE[seed]
    tr, te, ex = M[name]
    P, keep = res[name]
    if name == RULE_NAME:
        pred_ok = (P.sum(1) == 0) & keep
    else:
        pred_ok = (te == ok_cluster(tr, ex)) & keep
    return pred_ok, (E.sum(1) == 0), keep, (P == E).mean(1)

def met(pred_ok, exp_ok, keep, agree):
    TP = (pred_ok & exp_ok & keep).sum()
    FN = (~pred_ok & exp_ok & keep).sum()
    FP = (pred_ok & ~exp_ok & keep).sum()
    TN = (~pred_ok & ~exp_ok & keep).sum()
    n = TP + FN + FP + TN
    d = lambda a, b: a / b if b else np.nan
    acc = d(TP + TN, n)
    rec = d(TP, TP + FN)
    spe = d(TN, TN + FP)
    pre = d(TP, TP + FP)
    f1 = d(2 * pre * rec, pre + rec) if not (np.isnan(pre) or np.isnan(rec)) else np.nan
    pe = ((TP + FN) * (TP + FP) + (TN + FP) * (TN + FN)) / n**2 if n else np.nan
    kap = (acc - pe) / (1 - pe) if n and pe < 1 else np.nan
    return {
        'TP': TP, 'FN': FN, 'FP': FP, 'TN': TN,
        'Accuracy': acc, 'Sensitivity': rec, 'Specificity': spe,
        'Precision': pre, 'F1': f1, 'Kappa': kap,
        'Joint agreement': agree[keep].mean() if keep.any() else np.nan
    }

# Sanity check on Seed 42
p, e, k, a = bin_arrays('FCM + NSGA-II', 42)
print('Sanity check (FCM + NSGA-II, seed 42):')
print('mine  :', {k_: int(v) for k_, v in met(p, e, k, a).items() if k_ in ('TP', 'FN', 'FP', 'TN')})
print('yours :', {k_: metrics(STORE[42][1]['FCM + NSGA-II'][0], STORE[42][1]['FCM + NSGA-II'][1])[k_] for k_ in ('TP', 'FN', 'FP', 'TN')})

B, rng = 2000, np.random.default_rng(0)
names = [
    'KMeans', 'DBSCAN', 'Hierarchical', 'GMM', 'DBSCAN (matched retention)',
    'FCM', 'FCM + NSGA-II', 'fcm + grid search', RULE_NAME
]
keys = ['Accuracy', 'Sensitivity', 'Specificity', 'Precision', 'F1', 'Kappa', 'Joint agreement']

out = []
for name in names:
    arr = [bin_arrays(name, s) for s in RANDOM_SEEDS]
    pt = pd.DataFrame([met(*x) for x in arr])[keys].mean()
    boots = []
    for _ in range(B):
        idx = rng.integers(0, len(E), len(E))
        boots.append(pd.DataFrame([met(*(z[idx] for z in x)) for x in arr])[keys].mean())
    bt = pd.DataFrame(boots)
    lo, hi = bt.quantile(0.025), bt.quantile(0.975)
    out.append({'Method': name, **{k_: f'{pt[k_]:.3f} [{lo[k_]:.3f}, {hi[k_]:.3f}]' for k_ in keys}})

ci = pd.DataFrame(out).set_index('Method')
display(ci)
ci.to_csv('test_metrics_with_95CI.csv')
print("\nAll tables and figures have been successfully generated and saved.")
```

```python
# %%
# ==============================================================================
# 23. BOOTSTRAP 95% CONFIDENCE INTERVALS (OWN & MUTUAL SUBSETS - MEAN ± STD [CI])
# ==============================================================================
def get_eval_data(name, seed, use_common=False):
    (M, res) = STORE[seed]
    tr, te, ex = M[name]
    P, keep = res[name]
    
    # If evaluating on mutual subset, overwrite keep with the intersection of all keeps
    if use_common:
        keep = np.logical_and.reduce([r[1] for r in res.values()])
        
    if name == RULE_NAME:
        pred_ok = (P.sum(1) == 0) & keep
    else:
        pred_ok = (te == ok_cluster(tr, ex)) & keep
        
    return pred_ok, (E.sum(1) == 0), keep, (P == E).mean(1)

def met(pred_ok, exp_ok, keep, agree):
    TP = (pred_ok & exp_ok & keep).sum()
    FN = (~pred_ok & exp_ok & keep).sum()
    FP = (pred_ok & ~exp_ok & keep).sum()
    TN = (~pred_ok & ~exp_ok & keep).sum()
    n = TP + FN + FP + TN
    d = lambda a, b: a / b if b else np.nan
    acc = d(TP + TN, n)
    rec = d(TP, TP + FN)
    spe = d(TN, TN + FP)
    pre = d(TP, TP + FP)
    f1 = d(2 * pre * rec, pre + rec) if not (np.isnan(pre) or np.isnan(rec)) else np.nan
    pe = ((TP + FN) * (TP + FP) + (TN + FP) * (TN + FN)) / n**2 if n else np.nan
    kap = (acc - pe) / (1 - pe) if n and pe < 1 else np.nan
    
    return {
        'Retained': int(keep.sum()),
        'Joint agreement': agree[keep].mean() if keep.any() else np.nan,
        'Accuracy': acc, 
        'Precision': pre,
        'Sensitivity': rec, 
        'Specificity': spe,
        'F1': f1, 
        'κ': kap
    }

B, rng = 2000, np.random.default_rng(0)
names = [
    'KMeans', 'DBSCAN', 'Hierarchical', 'GMM', 'DBSCAN (matched retention)',
    'FCM', 'FCM + NSGA-II', 'fcm + grid search', RULE_NAME
]

# All keys including Retained for computation
all_keys = ['Retained', 'Joint agreement', 'Accuracy', 'Precision', 'Sensitivity', 'Specificity', 'F1', 'κ']
metric_keys = ['Joint agreement', 'Accuracy', 'Precision', 'Sensitivity', 'Specificity', 'F1', 'κ']

def compute_ci(use_common=False):
    total_samples = len(E)
    retained_col = f"Retained / {total_samples}"
    out = []
    
    for name in names:
        arr = [get_eval_data(name, s, use_common) for s in RANDOM_SEEDS]
        
        # Calculate point estimates: Mean and STD across the 5 seeds
        base_df = pd.DataFrame([met(*x) for x in arr])
        pt_mean = base_df[all_keys].mean()
        pt_std = base_df[all_keys].std(ddof=1).fillna(0.0)
        
        # Bootstrap over the test set to get 95% Confidence Intervals
        boots = []
        for _ in range(B):
            idx = rng.integers(0, total_samples, total_samples)
            boots.append(pd.DataFrame([met(*(z[idx] for z in x)) for x in arr])[all_keys].mean())
            
        bt = pd.DataFrame(boots)
        lo, hi = bt.quantile(0.025), bt.quantile(0.975)
        
        row = {'Method': name}
        
        # Format Retained (e.g., 53.0 ± 2.1 [45.0, 60.0])
        r_m, r_s, r_l, r_h = pt_mean['Retained'], pt_std['Retained'], lo['Retained'], hi['Retained']
        row[retained_col] = f"{r_m:.1f} ± {r_s:.1f} [{r_l:.1f}, {r_h:.1f}]"
        
        # Format metrics (e.g., 0.850 ± 0.012 [0.810, 0.890])
        for k_ in metric_keys:
            m, s, l, h = pt_mean[k_], pt_std[k_], lo[k_], hi[k_]
            row[k_] = f"{m:.3f} ± {s:.3f} [{l:.3f}, {h:.3f}]"
            
        out.append(row)
        
    cols = ['Method', retained_col] + metric_keys
    return pd.DataFrame(out)[cols].set_index('Method')

# 1. Calculate for OWN subset
print("\nComputing Bootstrapped CIs for OWN subset...")
ci_own = compute_ci(use_common=False)
display(ci_own)
ci_own.to_csv('test_metrics_with_95CI_own_subset.csv')

# 2. Calculate for MUTUAL subset
print("\nComputing Bootstrapped CIs for MUTUAL subset...")
ci_mutual = compute_ci(use_common=True)
display(ci_mutual)
ci_mutual.to_csv('test_metrics_with_95CI_mutual_subset.csv')

print("\nAll tables and figures have been successfully generated and saved.")
```
